```python
import pandas as pd
import numpy as np
import csv
```

```python
def load_candles(filepath):
    candles = []
    with open (filepath, "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            candle = {
                "date":row["Date"],
                "open":float(row["Open"]),
                "close":float(row["Close"]),
                "high":float(row["High"]),
                "low":float(row["Low"]),
            }
            candles.append(candle)
        return candles
```

```python
def calculate_ratios_for_one_candle(previous_candle, current_candle):
    curr_candle_close = current_candle["close"]
    curr_candle_high = current_candle["high"]
    curr_candle_low = current_candle["low"]

    prev_candle_close = previous_candle["close"]
    prev_candle_high = previous_candle["high"]
    prev_candle_low = previous_candle["low"]

    x1 = (curr_candle_close - prev_candle_close)/prev_candle_close
    x2 = (curr_candle_high - prev_candle_high)/prev_candle_high
    x3 = (curr_candle_low - prev_candle_low)/prev_candle_low
    x4 = (curr_candle_high - curr_candle_close)/curr_candle_close
    x5 = (curr_candle_close - curr_candle_low)/curr_candle_close

    return {
        "date":current_candle["date"],
        "x1_close_to_close":x1,
        "x2_high_to_high":x2,
        "x3_low_to_low":x3,
        "x4_high_to_close":x4,
        "x5_close_to_low":x5,
    }
```

```python
def calculate_ratios(candles):
    ratio_list = []

    for i in range(1, len(candles)):
        previous_candle = candles[i-1]
        current_candle = candles[i]

        ratios = calculate_ratios_for_one_candle(previous_candle, current_candle)
        ratio_list.append(ratios)
    return ratio_list
```

```python
filepath = "./sample_XAUUSD_D1_extended.csv"
candles = load_candles(filepath)
print(f"loaded {len(candles)}")
```

```python
ratios = calculate_ratios(candles)
print(len(ratios))
```

```python
for row in ratios[:5]:
    print(row)
```
